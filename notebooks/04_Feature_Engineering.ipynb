# Feature Engineering

## Objective

Feature Engineering is the process of preparing the dataset for machine learning models.

In this notebook, we will:

- Select input and output features
- Encode categorical variables
- Scale numerical features (if required)
- Create feature matrix (X) and target variable (y)
- Save the processed dataset

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/cleaned_insurance.csv")

df.head()
```

# Check Dataset

```python
df.info()
```

# Encode Categorical Features

Machine Learning models require numerical values.

We will convert the categorical columns into numerical values using Label Encoding.

```python
label_encoder = LabelEncoder()
```

```python
df["sex"] = label_encoder.fit_transform(df["sex"])

df["smoker"] = label_encoder.fit_transform(df["smoker"])

df["region"] = label_encoder.fit_transform(df["region"])

df.head()
```

# Verify Encoded Dataset

```python
df.info()
```

# Feature Selection

Separate the input features (X) and the target variable (y).

```python
X = df.drop("charges", axis=1)

y = df["charges"]
```

```python
print("Input Features (X)")
X.head()
```

```python
print("Target Variable (y)")
y.head()
```

# Feature Scaling

StandardScaler standardizes numerical features.

This step is useful for algorithms such as Linear Regression and Gradient Boosting.

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

X_scaled.head()
```

# Save Processed Dataset

The encoded dataset is saved for model building.

```python
encoded_df = X.copy()

encoded_df["charges"] = y
```

```python
encoded_df.to_csv("../data/encoded_insurance.csv", index=False)

print("Encoded dataset saved successfully!")
```
