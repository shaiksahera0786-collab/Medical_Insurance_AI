# Exploratory Data Analysis (EDA)

## Objective

Exploratory Data Analysis (EDA) helps us understand the dataset by identifying patterns, trends, relationships, and distributions.

In this notebook, we will perform:

- Univariate Analysis
- Bivariate Analysis
- Count Plot
- Histogram
- Box Plot
- Scatter Plot
- Pair Plot
- Correlation Matrix
- Heatmap

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/cleaned_insurance.csv")

df.head()
```

# Univariate Analysis

Univariate analysis examines one variable at a time to understand its distribution.

# Count Plot - Sex

This plot shows the number of male and female customers.

```python
plt.figure(figsize=(6,4))

sns.countplot(x="sex", data=df)

plt.title("Count Plot of Sex")

plt.xlabel("Sex")

plt.ylabel("Count")

plt.show()
```

# Count Plot - Smoker

This plot shows the number of smokers and non-smokers.

```python
plt.figure(figsize=(6,4))

sns.countplot(x="smoker", data=df)

plt.title("Count Plot of Smoker")

plt.xlabel("Smoker")

plt.ylabel("Count")

plt.show()
```

# Count Plot - Region

This plot shows the distribution of customers across different regions.

```python
plt.figure(figsize=(8,5))

sns.countplot(x="region", data=df)

plt.title("Count Plot of Region")

plt.xlabel("Region")

plt.ylabel("Count")

plt.show()
```

# Histogram - Age

Histogram shows the distribution of customer ages.

```python
plt.figure(figsize=(8,5))

sns.histplot(df["age"], bins=20, kde=True)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

# Histogram - BMI

Histogram shows the distribution of BMI values.

```python
plt.figure(figsize=(8,5))

sns.histplot(df["bmi"], bins=20, kde=True)

plt.title("BMI Distribution")

plt.xlabel("BMI")

plt.ylabel("Frequency")

plt.show()
```

# Box Plot - Age

The box plot helps identify the spread of the data and any outliers in the Age column.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["age"])

plt.title("Box Plot of Age")

plt.show()
```

# Box Plot - BMI

The box plot helps visualize the distribution and outliers in the BMI column.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["bmi"])

plt.title("Box Plot of BMI")

plt.show()
```

# Box Plot - Charges

The box plot shows the spread of medical insurance charges and highlights high-cost outliers.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["charges"])

plt.title("Box Plot of Charges")

plt.show()
```

# Bivariate Analysis

Bivariate analysis helps us understand the relationship between two variables.

# Scatter Plot - Age vs Charges

This plot shows how age influences medical insurance charges.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="age", y="charges", data=df)

plt.title("Age vs Charges")

plt.show()
```

# Scatter Plot - BMI vs Charges

This plot shows the relationship between BMI and medical insurance charges.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="bmi", y="charges", data=df)

plt.title("BMI vs Charges")

plt.show()
```

# Pair Plot

Pair Plot visualizes pairwise relationships among numerical features.

```python
sns.pairplot(df)

plt.show()
```

# Correlation Matrix

The correlation matrix measures the strength of relationships between numerical variables.

```python
df.corr(numeric_only=True)
```

# Heatmap

A heatmap provides a visual representation of the correlation matrix.

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```
