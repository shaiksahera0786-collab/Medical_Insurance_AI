# Save Best Model

## Objective

In this notebook, we will save the best-performing machine learning model using Joblib.

The saved model will be used later in the Streamlit web application for making predictions.

```python
import joblib
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
```

```python
df = pd.read_csv("../data/encoded_insurance.csv")

df.head()
```

```python
X = df.drop("charges", axis=1)

y = df["charges"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

# Train Final Model

Train the best-performing model using the entire training dataset.

```python
model = GradientBoostingRegressor(
    learning_rate=0.05,
    max_depth=3,
    n_estimators=100,
    subsample=0.9,
    random_state=42
)
```

```python
model.fit(X_train, y_train)

print("Final Model Trained Successfully!")
```

```python
import joblib

joblib.dump(model, "../models/insurance_model.pkl")

print("Model saved successfully!")
```

```python
loaded_model = joblib.load("../models/insurance_model.pkl")

print("Model loaded successfully!")
```

```python
sample_data = X.iloc[[0]]

prediction = loaded_model.predict(sample_data)

print("Predicted Insurance Charges:", prediction[0])
```
