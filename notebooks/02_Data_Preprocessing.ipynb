# Data Preprocessing

## Objective

Data preprocessing is an important step in Machine Learning. It helps improve the quality of the dataset before training the models.

In this notebook, we will:

- Check missing values
- Handle missing values
- Check duplicate values
- Remove duplicate values
- Detect outliers
- Handle outliers (if necessary)
- Save the cleaned dataset

# Import Libraries

```python
import pandas as pd
import numpy as np
```

# Load Dataset

```python
df = pd.read_csv("../data/insurance.csv")

df.head()
```

# Check Missing Values

Missing values can affect the performance of machine learning models.

We will check whether the dataset contains any missing values.

```python
df.isnull().sum()
```

# Handle Missing Values

If missing values are present, they should be handled appropriately.

Since the Insurance dataset has no missing values, no action is required.

```python
df.dropna(inplace=True)

df.isnull().sum()
```

# Check Duplicate Values

Duplicate records can negatively affect model performance.

We will identify duplicate rows in the dataset.

```python
df.drop_duplicates(inplace=True)

df.duplicated().sum()
```

```python
print("Dataset Shape After Removing Duplicates:", df.shape)
```

# Check Outliers

Outliers are extreme values that are significantly different from other observations.

We will use Box Plots to identify outliers in the numerical columns.

## Import Visualization Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## Box Plot for Age

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["age"])
plt.title("Box Plot of Age")
plt.show()
```

## Box Plot for BMI

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["bmi"])
plt.title("Box Plot of BMI")
plt.show()
```

## Box Plot for Children

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["children"])
plt.title("Box Plot of Children")
plt.show()
```

## Box Plot for Charges

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["charges"])
plt.title("Box Plot of Charges")
plt.show()
```

# Handle Outliers

After visualizing the box plots, we observe that the `charges` column contains some outliers.

These values represent customers with very high medical expenses and are genuine observations, not data entry errors.

Therefore, we will **not remove the outliers** because they are important for building an accurate insurance cost prediction model.

## Verify Dataset Shape

```python
print("Dataset Shape:", df.shape)
```

# Save Cleaned Dataset

The cleaned dataset is saved for use in the next stages of the project.

```python
df.to_csv("../data/cleaned_insurance.csv", index=False)

print("Cleaned dataset saved successfully!")
```
