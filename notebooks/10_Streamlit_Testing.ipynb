# Streamlit Testing

## Objective

This notebook tests the saved machine learning model by loading it and making predictions using sample input values.

This confirms that the model is ready to be integrated into the Streamlit web application.

```python
import joblib
import pandas as pd
```

```python
model = joblib.load("../models/insurance_model.pkl")

print("Model Loaded Successfully!")
```

```python
sample_data = pd.DataFrame({
    "age": [30],
    "sex": [1],
    "bmi": [28.5],
    "children": [2],
    "smoker": [0],
    "region": [2]
})
```

```python
prediction = model.predict(sample_data)

print("Predicted Insurance Charges:", prediction[0])
```

```python
sample_data2 = pd.DataFrame({
    "age": [45],
    "sex": [0],
    "bmi": [35.2],
    "children": [3],
    "smoker": [1],
    "region": [1]
})
```

```python
prediction2 = model.predict(sample_data2)

print("Predicted Insurance Charges:", prediction2[0])
```

# Conclusion

The saved Gradient Boosting model was successfully loaded and tested with sample inputs.

The model generated predictions without any errors, confirming that it is ready to be integrated into a Streamlit web application.
